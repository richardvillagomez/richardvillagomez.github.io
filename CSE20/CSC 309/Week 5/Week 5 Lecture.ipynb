## Debugging

Robert writes the following code to calculate the division of two integer numbers. The code takes two integer numbers from a user, perform integer division, obtain the quotient and the remainder, and print out the results. There are several errors in this code. Can you help Robert find them *without running the code*?

````python
int_1 = input("Enter your first integer:")
int_2 = input("Enter your second integer:")

# casting int_1 and int_2 as integers so we can do arithmetic
int_1 = int(int_1)
int_2 = int(int_2)

quotient = int_1 / int_2
remainder = int_1 % int_2
print(f"The quotient is {quotient} and the remainder is {remainder}.")
````

## Pair programming
Let's write a program to convert a numerical grade into a letter grade. The program first prompts the user for a numerical grade between 0 to 100. Then, converts it into a letter grade based on the table. Finally, print a message to display the letter grade.

- Between 90 to 100: A
- Between 80 to 90: B
- Between 70 to 80: C
- Between 60 to 70: D
- Below 60: F



````python
score = input("Enter your numerical grade:")
score = int(score)


if score >= 90 and score <= 100:
  letter = "A"

elif score >= 80 and score < 90:
  letter = "B"

elif score >= 70 and score < 80:
  letter = "C"

elif score >= 60 and score < 70:
  letter = "D"

else:
  letter = "F"

print(f"Your final letter grade is {letter}.")
````
