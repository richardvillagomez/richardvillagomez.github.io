## Question 1: Voting
Let's write a program to decide whether a person is old enough to vote or not. The program takes a person's age as input, and prints it out (e.g., `"Your age is 20."`). Then, if the person is at least 18 years old, the program prints a message `"You are old enough to vote!"`.

## Question 2: Pass or not?
Write a program to inform whether a student passes a class or not. The overall grade of a student is calculated based on three weighted exams. Each exam is graded out of 100 points, and the weights of the three exams are: 30%, 30%, and 40%. A student will pass this class if they get 60 or higher for their overall grade. Otherwise, they fail the class. The program takes the grades of the three exams as input from a user, calculates the overall grade and prints out a message on whether the student passes or not.  

Below are two sample runs of the code.

### Sample run 1:

```
Enter score on Exam 1 (out of 100): 100  

Enter score on Exam 2 (out of 100): 100  

Enter score on Exam 3 (out of 100): 80

Your overall grade is 92.0.  Congratulation! You pass the class!
```
### Sample run 2:

```
Enter score on Exam 1 (out of 100): 50  

Enter score on Exam 2 (out of 100): 60  

Enter score on Exam 3 (out of 100): 60

Your overall grade is 57.0.  Sorry, you fail the class.
```

## Question 3: Tweet Decoder
I am not quite familiar with those common abbreviations in online communication such as messages in Twitter or WhatsApp. Could you write a program to help me decode them? The program will ask a user to enter one of the following abbreviations and then prints the corresponding English phrase.

LOL → Laughing out loud  
B4N → Bye for now  
DM  → Direct message  
BTW → By the way

If the user enters something else, the program prints `Sorry, don't know that one`.
