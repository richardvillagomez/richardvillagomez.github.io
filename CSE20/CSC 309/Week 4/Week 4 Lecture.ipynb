## Booleans and Comparison
The [boolean](https://docs.python.org/3/c-api/bool.html) type has exactly two values: `True` and `False`. We can use **comparison operators** to compare two values and output a boolean accordingly. Let's go through them one-by-one.

### 1.) equal: `==`
The **equal** operator outputs `True` if the two values are equal, and `False` otherwise.


```python
# set name to Michelle
name = 'Michelle'
```

```python
# check whether name is equal to Michelle
name == 'Michelle'
```

```python
# check whether name is equal to Mitchell
name == 'Mitchell'
```

We can also use the equal operator to compare integers or floats!

```python
# set height to 68.0 (unit is inches)
height = 68.0
```

```python
height == 68
```

```python
height == 68.1
```

### 2.) not equal: `!=`
The **not equal** operator outputs True if the two values are _not_ equal, and False otherwise.

```python
# set pet to dog
pet = 'dog'
```

```python
# check whether pet is not equal to wolf
pet != 'wolf'
```

```python
# check whether pet is not equal to bulldog
pet != 'bulldog'
```

### 3.) less than: `<`
The less than operator outputs `True` if the left value is _less than_ the right value, and `False` otherwise.

```python
# set salary to 50000
salary = 50000
```

```python
# check whether salary is less than 75000
salary < 75000
```

```python
# check whether salary is less than 50000
salary < 50000
```

### 4.) less than or equal to: `<=`
The less than or equal to operator outputs `True` if the left value is _less than or equal to_ the right value, and `False` otherwise.


```python
# check whether salary is less than or equal to 50000
salary <= 50000
```

### 5.) greater than: `>`
The greater than operator outputs `True` if the left value is _greater than_ the right value, and `False` otherwise.

```python
salary > 49999
```

```python
salary > 500000
```

### 6.) greater than or equal to: `!=`
The **greater than or equal to** operator outputs `True` if the left value is greater than or equal to the right value, and `False` otherwise.


```python
salary >= 50000
```

## Conditionals
In addition, we can use **conditionals** to check whether a _condition_ is `True` or `False`. Depending on the condition, we might want to take a different course of action. Let's go through the different kinds of conditionals.

### 1.) `if` statement
An `if` statement lets us execute certain code _if_ a condition is true.

```python
# set age
age = 45

# initialize drivers_license_eligible
drivers_license_eligible = False
```

```python
# if age is at least 16, set drivers_license_eligible to True
if age >= 16:
    drivers_license_eligible = True
```

```python
print(drivers_license_eligible)
```

### 2.) `elif` statement
An `elif` statement lets us handle additional branches. In other words, if there are other conditions we want to check for, we can do so using `elif` statements.


```python
# set age to 15 and a half
age = 15 + 1/2
print(age)

# initialize drivers_permit_eligible
drivers_permit_eligible = False
```

```python
if age >= 16:
    drivers_license_eligible = True
    drivers_permit_eligible = True
elif age >= 15.5:
    drivers_permit_eligible = True
```

```python
print(drivers_permit_eligible)
```

### 3.) `else` statement
An `else` statement lets us do something in the case that all the other cases failed.

```python
age = 15

if age >= 16:
    drivers_license_eligible = True
    drivers_permit_eligible = True
elif age >= 15.5:
    drivers_permit_eligible = True
else:
    print("Sorry, you're not eligible to drive!")
```
