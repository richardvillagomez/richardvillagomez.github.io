## PRIMM
PRIMM is an approach of structuring programming lessons. It stands for Predict, Run, Investigate, Modify and Make, representing different stages of a lesson, or series of lessons.

PRIMM focuses on:
- **Reading code** before writing code
- **Working collaboratively** to talk about programs
- Reducing cognitive load
- **Gradually taking ownership** of programs when ready

There are five stages of PRIMM:
1. **Predict**: Students discuss a program and predict what it might do. The focus is on the function of the code.
2. **Run**: Students run the program and test their prediction
3. **Investigate**: Teacher provides a range of activities to explore the code, such as tracing, explaining, debugging, etc.
4. **Modify**: Students edit a program to change its functionality via a sequence of increasingly more challenging exercises
5. **Make**: Students create a new program that uses the same structures
