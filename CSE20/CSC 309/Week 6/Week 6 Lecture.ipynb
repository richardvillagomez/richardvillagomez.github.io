## `range()`
`range(<start>, <end>)` generates a list of numbers from `<start>` to `<end>-1` (i.e., excluding `<end>`).

For example, `range(1, 10)` generates numbers 1, 2, 3, ..., 9:

````python
values = range(1, 10)

# print all values one-by-one
for value in values:
  print(value)
````

---
## `for` loops

`for` loop (combined with range()) can repeatedly run a block of code.
The syntax of a `for` loop is like this:

```
for variable in range(<start>, <end>): # The colon is mandatory
     ------------------------------------------------------
    | Block of code to execute repeatedly                   |
    | Note: Must be indented                                |
     ------------------------------------------------------
```

As for each number generated by `range()`, the number is assigned to variable and the code in the loop body is executed with that number.
