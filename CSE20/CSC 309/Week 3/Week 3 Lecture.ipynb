## Variables
In natural language, we have terminology that lets us quickly reference very complicated concepts. We don't say, "That's a large mammal with white fur and sharp teeth!" Instead, we just say, "Polar bear!"

In Python, we do this with assignment statements. An assignment statement has a name on the left side of an `=` sign and an expression to be evaluated on the right.

````python
pet = 'polar bear'
````

Once we've defined a variable, we can use its name to refer to its value. Notice that `print(pet)` generates the same output as `print('polar bear'`):

````python
print(pet)
````

````python
print('polar bear')
````

We can also change the value stored in a variable by **reassigning it.** In other words, we are assigning the variable to hold  different value.

````python
pet = 'dog'
print(pet)
````

We can use a variable when assigning other variables, and we can even use a variable's current value to reassign itself:

````python
pet = 'small ' + pet
print(pet)
````

## Naming
In Python, variable names must begin with a letter or underscore, and may contain only letters, numbers, and underscores. Common convention is to name your variables in **`lower_snake_case`**: lower-case letters with underscores between words, if needed.

In addition, make sure to use descriptive names for variables to ensure your code is **readable** and **interpretable**. Watch out for keywords and built-in things like `class` and `print`; you can't use these as variable names.

## Type conversions
So far, we've learned about a few different **types** in Python. As a refresher:
- Strings (str) are blocks of text, surrounded by single quotes, double quotes, or triple quotes.
- Integers (int) are whole numbers.
- Floats (float): numbers with decimals.

Python allows for **type conversions**, so that we can convert data from one type to another. Specifically:
- To convert a **float** number to an **int** number, use `int()`.
- To convert an **int** number to a **float** number, use `float()`.
- To convert any data to a string, use `str()`.

````python
new_int = int(3.0)
print(new_int)
````

````python
type(new_int)
````

````python
new_float = float(3)
print(new_float)
````

````python
type(new_float)
````

````python
new_str = str(3)
print(new_str)
````

````python
type(new_str)
````

````python
new_str_2 = str(3.0)
print(new_str_2)
````

## f-strings
There is one more variation of strings in Python. [**f-strings**](https://www.geeksforgeeks.org/python/formatted-string-literals-f-strings-python/) (short for format string literals) are strings that make it very convenient to include Python names and complex expressions in strings, without having to worry about how many spaces/newlines separate input.
> f-strings are strings prefixed with f. Inside the string, expressions can be written within curly braces. When f-strings are evaluated, they evaluate any expressions within them and then create the corresponding string. Read more in the official Python f-string documentation.

Run the cell below to see an example.

````python
# just run this cell
animals = "foxes"
animal_count = 3
animal_fstr = f"There are {animal_count} {animals} in the farm."
animal_fstr
````

## More operations!
There are two more operators that will come in handy: **integer division** and **modulus**.

Integer division is written `//` and returns the result of a division (i.e., quotient) without any remainder. For instance, `11 // 5` is 2, since 5 goes into 11 two times, with some remainder.

````python
print(11 // 5)
````

The other operator is called modulus. It's kind of the opposite of integer division. It's written with `%`, and returns the remainder of the left number divided by the right. So, for instance, `11 % 5` is 1, since 11 divided by 5 is 2 with a remainder of 1.


````python
print(11 % 5)
````
